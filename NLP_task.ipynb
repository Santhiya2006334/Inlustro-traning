```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk.corpus import stopwords
nltk.download('punkt') #Download tokenizer
nltk.download('punkt_tab')
nltk.download('wordnet') #Download WordNet for Lemmatization
nltk.download('omw-1.4') # WordNet dependency
nltk.download('stopwords') #Download stopwords
```

```python
text="The beautiful ground in the earth"
tokens = word_tokenize(text)
print("Quotes:",tokens)
```

```python
porter=PorterStemmer()
stemmed_words=[porter.stem(word) for word in tokens]
print("Porter Stemmer:",stemmed_words)
```

```python
lemmatizer=WordNetLemmatizer()
lemmatized_words=[lemmatizer.lemmatize(word)for word in tokens]
print("Lemmatized Word:",lemmatized_words)
```

```python
tokens=[word.lower()for word in tokens]
print("Lowercased Tokens:",tokens)
```

```python
default_stopwords=set(stopwords.words('english'))
default_filtered_tokens=[word for word in tokens if word not in default_stopwords]
print("Tokens after removing default stopword:",default_filtered_tokens)
```

```python
#removing customer stopword
Custom_stopword={"bats",","}
custom_filtered_tokens=[word for word in default_filtered_tokens if word not in Custom_stopword]
print("Tokens after removing custom stopword:", custom_filtered_tokens)
```

```python
from nltk import pos_tag
import nltk
nltk.download('averaged_perceptron_tagger_eng')
pos_tags=pos_tag(custom_filtered_tokens)
print("Pos Tags:",pos_tags)
```
